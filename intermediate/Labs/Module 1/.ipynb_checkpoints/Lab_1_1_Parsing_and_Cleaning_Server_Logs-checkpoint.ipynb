A dirty dataset fed into even the best ML model produces unreliable predictions. Data cleaning is not boring prep work — it is **engineering judgment encoded as code**.

**`ModuleNotFoundError: No module named '...'`**
> Package is missing from the active Python environment.
> Fix: Run the pip install command above in a terminal, then **restart the kernel**.

**`CalledProcessError` — `--break-system-packages` / exit status 2**
> You are using a virtual environment (e.g. `myenv`) where that flag is not supported, or your pip version is old.
> Fix: Open a terminal, activate your venv (`source myenv/bin/activate`), then run `pip install <package>` without that flag.

**`Failed building wheel for <package>` / C extension errors**
> The package does not support your Python version (most common on Python 3.14).
> Fix: Switch the kernel to **Python 3.13**. Click the kernel name in the VS Code top-right corner → *Select Another Kernel* → *Python 3.13*. Then re-run.

> Fix: Check the kernel shown in the top-right of VS Code. Open a terminal, activate that environment, and install packages there.

**`PermissionError` or `[Errno 13]` when installing**
> Trying to install into a read-only system Python.
> Fix: Use a virtual environment — `python -m venv myenv && source myenv/bin/activate` — then install.

## Step 1: Generate Synthetic Messy Server Log Data

In a real engagement you'd load data from a SIEM, an ELK stack, or a Nutanix Prism Central export. For this lab we generate synthetic data that mirrors the types of problems you'll encounter in production.

We deliberately inject the following problems:

| Problem | How it gets injected |
|---------|---------------------|
| Malformed timestamps | Replace some valid timestamps with garbage strings |
| Missing host names | Set ~10% of `host` values to `None` |
| Mixed-case / typo log levels | Use a noisy vocabulary including `"EROR"`, `"warining"` |
| Out-of-range CPU | Allow values up to 115 |
| Negative memory | Allow negative values for ~5% of rows |
| Extreme response time outliers | Inject a handful of 10,000+ ms values |
| Random ~15% NaN across numeric columns | Simulate dropped telemetry |

> **Instructor Note:** Emphasize that generating synthetic data is itself a useful skill — it lets you test your pipeline before you get access to real (often sensitive) production data.

````python
def generate_messy_logs(n_rows=500, seed=42):
    """
    Generate a DataFrame simulating raw, messy Nutanix server logs.
    Intentionally introduces data quality issues for cleaning exercises.
    """
    rng = np.random.default_rng(seed)

    # --- Timestamps ---
    # Generate a sequence of valid timestamps over 7 days
    base_time = pd.Timestamp('2024-03-01 00:00:00')
    time_offsets = pd.to_timedelta(rng.integers(0, 7 * 24 * 3600, n_rows), unit='s')
    timestamps = [(base_time + offset).strftime('%Y-%m-%d %H:%M:%S') for offset in time_offsets]

    # Inject ~8% malformed timestamps
    bad_ts_values = ['N/A', '2024-99-01 00:00:00', 'null', '--', '03/01/2024 12:00']
    bad_indices = rng.choice(n_rows, size=int(n_rows * 0.08), replace=False)
    for idx in bad_indices:
        timestamps[idx] = rng.choice(bad_ts_values)

    # --- Host names ---
    # Realistic Nutanix CVM and AHV host naming conventions
    hosts = [
        f'ntnx-cvm-{str(i).zfill(3)}' for i in range(1, 11)
    ] + [
        f'ntnx-ahv-{str(i).zfill(3)}' for i in range(1, 6)
    ]
    host_col = rng.choice(hosts, n_rows).tolist()

    # Inject ~10% missing hosts
    missing_host_idx = rng.choice(n_rows, size=int(n_rows * 0.10), replace=False)
    for idx in missing_host_idx:
        host_col[idx] = None

    # --- Log levels ---
    # Clean values mixed with common real-world typos and case variants
    clean_levels   = ['INFO', 'WARNING', 'ERROR', 'DEBUG', 'CRITICAL']
    noisy_levels   = ['info', 'Info', 'EROR', 'warining', 'WARNING', 'Warng',
                      'ERROR', 'error', 'DEBUG', 'debug', 'CRITICAL', 'critical',
                      'INFO', 'INFO', 'INFO', 'WARNING', 'WARNING']  # weight toward INFO/WARNING
    log_level_col  = rng.choice(noisy_levels, n_rows).tolist()

    # --- Components ---
    components = ['AOS', 'Prism', 'Stargate', 'Cerebro', 'Curator', 'Acropolis',
                  ' Stargate', 'prism', 'CURATOR']  # include whitespace & case issues
    component_col = rng.choice(components, n_rows).tolist()

    # --- CPU percent ---
    # Normal range 5-95, but allow out-of-range values
    cpu = rng.uniform(5, 95, n_rows)
    # Inject ~3% out-of-range (> 100)
    oor_cpu_idx = rng.choice(n_rows, size=int(n_rows * 0.03), replace=False)
    cpu[oor_cpu_idx] = rng.uniform(101, 115, len(oor_cpu_idx))

    # --- Memory (MB) ---
    memory = rng.uniform(512, 32768, n_rows)
    # Inject ~5% negative values (sensor read errors)
    neg_mem_idx = rng.choice(n_rows, size=int(n_rows * 0.05), replace=False)
    memory[neg_mem_idx] = rng.uniform(-500, -1, len(neg_mem_idx))

    # --- Disk I/O (MB/s) ---
    disk_io = rng.uniform(0, 500, n_rows)

    # --- Response time (ms) ---
    response_time = rng.exponential(scale=150, size=n_rows)  # skewed right — realistic
    # Inject extreme outliers (~1%)
    outlier_rt_idx = rng.choice(n_rows, size=int(n_rows * 0.01), replace=False)
    response_time[outlier_rt_idx] = rng.uniform(10000, 60000, len(outlier_rt_idx))

    # --- Error codes ---
    error_codes = ['ERR-1001', 'ERR-2042', 'ERR-3300', 'NX-4401', 'NX-5500',
                   'ERR-1234', 'NX-9900', None, None, None]  # 30% missing
    error_code_col = rng.choice(error_codes, n_rows).tolist()

    # --- Log messages ---
    # Realistic Nutanix log message patterns
    message_templates = [
        'Stargate: WRITE op completed in {rt:.0f}ms [ERR-1001] on vdisk-{vid}',
        'AOS: READ request from guest vm-{vid} latency={rt:.0f}ms',
        'Cerebro: Snapshot DELETE failed [NX-4401] on container-{vid}',
        'Prism: WRITE throughput degraded, disk_io={io:.1f} MB/s [ERR-2042]',
        'Curator: Scheduled READ scan completed on node ntnx-cvm-{cvm:03d}',
        'Acropolis: VM migration WRITE phase started for vm-{vid}',
        'Stargate: DELETE operation [NX-5500] vdisk reclaim triggered',
        'AOS: Heartbeat READ OK, cpu={cpu:.1f}% mem={mem:.0f}MB',
        'Prism: Configuration UPDATE applied [ERR-3300] cluster-wide',
        'Cerebro: WRITE replication lag detected on remote site',
        'Curator: READ-REPAIR initiated for EC strip [NX-9900]',
        'Acropolis: WRITE IOPS limit reached on vm-{vid}, throttling applied',
        'AOS: DELETE of stale extent group [ERR-1234] completed',
        'Stargate: READ cache miss ratio={rt:.0f}% above threshold'
    ]
    messages = [
        rng.choice(message_templates).format(
            rt=rng.uniform(10, 5000),
            vid=rng.integers(100, 999),
            io=rng.uniform(0, 600),
            cvm=rng.integers(1, 10),
            cpu=rng.uniform(0, 100),
            mem=rng.uniform(512, 32768)
        )
        for _ in range(n_rows)
    ]

    # --- Assemble the raw DataFrame ---
    df = pd.DataFrame({
        'timestamp':       timestamps,
        'host':            host_col,
        'log_level':       log_level_col,
        'component':       component_col,
        'message':         messages,
        'cpu_percent':     cpu,
        'memory_mb':       memory,
        'disk_io_mbps':    disk_io,
        'response_time_ms': response_time,
        'error_code':      error_code_col
    })

    # --- Inject ~15% random NaN into numeric columns ---
    numeric_cols = ['cpu_percent', 'memory_mb', 'disk_io_mbps', 'response_time_ms']
    for col in numeric_cols:
        nan_idx = rng.choice(n_rows, size=int(n_rows * 0.15), replace=False)
        df.loc[nan_idx, col] = np.nan

    return df


# Generate the data
raw_df = generate_messy_logs(n_rows=500, seed=RANDOM_SEED)

# Save to CSV so we always have the original raw file
raw_df.to_csv('raw_server_logs.csv', index=False)
print(f"Raw log data saved to raw_server_logs.csv")
print(f"Shape: {raw_df.shape[0]} rows x {raw_df.shape[1]} columns")
print()

# Preview the first 10 rows
raw_df.head(10)
````

## Step 3: Fix Timestamps

The `timestamp` column is stored as a plain string. Some entries are valid ISO 8601 strings; others are garbage. We need to:

1. Parse valid timestamps into proper `datetime` objects
2. Identify and handle the ones that cannot be parsed
3. Extract useful time-based features (hour of day, day of week) — these often have strong predictive signal in infrastructure anomaly detection

**Key parameter:** `pd.to_datetime(errors='coerce')` — instead of raising an exception on bad input, it silently converts unparseable strings to `NaT` (Not a Time). This lets us identify and handle them in a controlled way.

> **Instructor Note:** Ask the class: *"We have rows with unparseable timestamps. Should we drop them or fill them? What are the trade-offs?"* Dropping is cleaner but wastes data. Forward-fill assumes the previous timestamp is close — risky if rows are out of order. This is a design decision that should be documented in your pipeline.

````python
print("--- Before parsing ---")
print(f"dtype: {df['timestamp'].dtype}")
print("Sample values:")
print(df['timestamp'].head(8).to_string())

# Step 3a: Parse with coerce — invalid strings become NaT
df['timestamp'] = pd.to_datetime(df['timestamp'], errors='coerce')

# Step 3b: Count how many failed to parse
n_nat = df['timestamp'].isna().sum()
print(f"\n--- After pd.to_datetime(errors='coerce') ---")
print(f"dtype: {df['timestamp'].dtype}")
print(f"Failed to parse (NaT): {n_nat} rows ({n_nat/len(df)*100:.1f}%)")
````

````python
# Step 3c: Decision — drop rows with unparseable timestamps
# Rationale: we cannot assign events to a time window for time-series modeling
# without a valid timestamp. Filling with a fake time would introduce false signal.

rows_before = len(df)
df = df.dropna(subset=['timestamp']).reset_index(drop=True)
rows_after  = len(df)

print(f"Dropped {rows_before - rows_after} rows with NaT timestamps.")
print(f"Remaining rows: {rows_after}")

# --- Alternative approach for instructor discussion ---
# df['timestamp'] = df['timestamp'].fillna(method='ffill')  # forward-fill
# This only makes sense if rows are sorted by time AND gaps are small.
# Always document the choice and why!
````

````python
# Step 3d: Extract time-based features
# These become model features — infrastructure load follows daily/weekly patterns

df['hour_of_day']  = df['timestamp'].dt.hour          # 0-23
df['day_of_week']  = df['timestamp'].dt.day_name()    # Monday, Tuesday, ...
df['date']         = df['timestamp'].dt.date          # For grouping / daily aggregation

print("New time-based features added:")
print(df[['timestamp', 'hour_of_day', 'day_of_week', 'date']].head(8).to_string())

# Quick sanity check: distribution of events by hour
fig, ax = plt.subplots(figsize=(10, 3))
df['hour_of_day'].value_counts().sort_index().plot(kind='bar', ax=ax, color='steelblue')
ax.set_title('Log Events by Hour of Day')
ax.set_xlabel('Hour')
ax.set_ylabel('Event Count')
plt.tight_layout()
plt.show()
````

## Step 4: Standardize Categorical Columns

Categorical columns like `log_level` and `component` are deceptively tricky. A model will treat `"ERROR"`, `"error"`, `"EROR"`, and `"ERR"` as four completely different categories unless you standardize them first.

**Our strategy:**
1. Strip leading/trailing whitespace (a silent killer — `" Stargate"` != `"Stargate"`)
2. Normalize case
3. Apply an explicit typo-correction mapping — this requires domain knowledge

> **Instructor Note:** The typo-correction map is business logic. It belongs in your code, not in a data dictionary hidden in a spreadsheet. Ask: *"What happens if we skip step 3 and just do `.upper()`? What happens to `'EROR'`?"*

````python
# --- Fix log_level ---

print("BEFORE cleaning:")
print(df['log_level'].value_counts().to_string())

# Step 1: Strip whitespace and uppercase everything
df['log_level'] = df['log_level'].str.strip().str.upper()

# Step 2: Map typos to canonical values
# This mapping is intentionally explicit — no magic regex guessing
log_level_typo_map = {
    'EROR':    'ERROR',
    'ERR':     'ERROR',
    'WARINING': 'WARNING',
    'WARN':    'WARNING',
    'WARNG':   'WARNING',
    'WRNG':    'WARNING',
    'DBG':     'DEBUG',
    'CRIT':    'CRITICAL',
    'INFO':    'INFO',     # already correct, included for completeness
}
df['log_level'] = df['log_level'].replace(log_level_typo_map)

# Step 3: Any value not in the expected set gets labelled UNKNOWN
valid_levels = {'INFO', 'WARNING', 'ERROR', 'DEBUG', 'CRITICAL'}
df['log_level'] = df['log_level'].where(df['log_level'].isin(valid_levels), other='UNKNOWN')

print()
print("AFTER cleaning:")
print(df['log_level'].value_counts().to_string())
````

````python
# --- Fix component ---

print("BEFORE cleaning:")
print(df['component'].value_counts().to_string())

# Strip whitespace, then apply title case
# 'AOS' is an acronym — we handle it explicitly rather than letting title case mangle it
df['component'] = df['component'].str.strip().str.title()

# Title case will turn 'AOS' into 'Aos' — fix acronyms back
component_fix_map = {
    'Aos': 'AOS',
}
df['component'] = df['component'].replace(component_fix_map)

print()
print("AFTER cleaning:")
print(df['component'].value_counts().to_string())
````

````python
# Visualize the cleaned categoricals side by side
fig, axes = plt.subplots(1, 2, figsize=(14, 4))

# Log level distribution
level_order = ['INFO', 'WARNING', 'ERROR', 'DEBUG', 'CRITICAL', 'UNKNOWN']
level_colors = {'INFO': '#4caf50', 'WARNING': '#ff9800', 'ERROR': '#f44336',
                'DEBUG': '#2196f3', 'CRITICAL': '#9c27b0', 'UNKNOWN': '#9e9e9e'}

counts = df['log_level'].value_counts().reindex(level_order, fill_value=0)
bars = axes[0].bar(counts.index, counts.values,
                   color=[level_colors.get(l, 'gray') for l in counts.index])
axes[0].set_title('Log Level Distribution (Cleaned)')
axes[0].set_ylabel('Count')
axes[0].tick_params(axis='x', rotation=15)
for bar, val in zip(bars, counts.values):
    axes[0].text(bar.get_x() + bar.get_width()/2, bar.get_height() + 2,
                 str(val), ha='center', va='bottom', fontsize=9)

# Component distribution
comp_counts = df['component'].value_counts()
axes[1].barh(comp_counts.index, comp_counts.values, color='steelblue')
axes[1].set_title('Events by Component (Cleaned)')
axes[1].set_xlabel('Count')

plt.tight_layout()
plt.show()
````

## Step 5: Handle Missing Numeric Values

Missing numeric values require a deliberate strategy. The choices are:

| Strategy | When to use |
|----------|-------------|
| **Drop the row** | When < 5% missing and the remaining data is sufficient |
| **Fill with mean** | When data is normally distributed and missings are random |
| **Fill with median** | When data is skewed (outliers distort the mean) — **preferred for telemetry** |
| **Fill with 0 / domain constant** | When "missing" has a known domain meaning |
| **Model-based imputation** | When patterns in other columns predict the missing value |
| **Flag as a feature** | When "missing" itself is a signal (sensor failure = problem!) |

> **Instructor Note:** For infrastructure data, missing CPU/memory values are often NOT missing at random — they frequently coincide with the very events you're trying to predict (node overload, agent crash). A `is_cpu_missing` binary feature can itself be predictive. We'll cover this in Module 3.

````python
# First: ensure numeric columns have the right dtype
# (they may still be 'object' from the CSV load if they had mixed content)
numeric_cols = ['cpu_percent', 'memory_mb', 'disk_io_mbps', 'response_time_ms']

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Show missing % per column before imputation
print("Missing values BEFORE imputation:")
missing_before = df[numeric_cols].isnull().sum()
for col in numeric_cols:
    pct = df[col].isnull().mean() * 100
    print(f"  {col:<22} {df[col].isnull().sum():>4} missing ({pct:.1f}%)")
````

````python
# --- Optional: flag missingness before filling ---
# Uncomment these lines in a real project — the flag itself can be a useful feature
# df['cpu_missing']      = df['cpu_percent'].isnull().astype(int)
# df['response_missing'] = df['response_time_ms'].isnull().astype(int)

# cpu_percent: fill with median
# Response time for cpu is skewed by high-load spikes — median is robust
cpu_median = df['cpu_percent'].median()
df['cpu_percent'] = df['cpu_percent'].fillna(cpu_median)
print(f"cpu_percent   median fill value: {cpu_median:.2f}%")

# memory_mb: fill with median
mem_median = df['memory_mb'].median()
df['memory_mb'] = df['memory_mb'].fillna(mem_median)
print(f"memory_mb     median fill value: {mem_median:.0f} MB")

# disk_io_mbps: fill with 0
# Domain knowledge: a missing I/O reading most likely means the node was idle
# (agent only reports when activity is above threshold on many Nutanix deployments)
df['disk_io_mbps'] = df['disk_io_mbps'].fillna(0.0)
print(f"disk_io_mbps  filled with: 0.0 (domain: idle node assumption)")

# response_time_ms: fill with median
rt_median = df['response_time_ms'].median()
df['response_time_ms'] = df['response_time_ms'].fillna(rt_median)
print(f"response_time_ms median fill value: {rt_median:.1f} ms")

print()
print("Missing values AFTER imputation:")
for col in numeric_cols:
    print(f"  {col:<22} {df[col].isnull().sum():>4} missing")
````

## Step 6: Fix Out-of-Range and Invalid Values

Even after handling `NaN`s, we have physically impossible values:

- `cpu_percent = 110.3` — a percentage cannot exceed 100
- `memory_mb = -342` — memory cannot be negative
- `response_time_ms = 48,000` — a 48-second response time for an internal Stargate call is an instrumentation error, not a real event

These outliers will corrupt regression models and distort distance-based algorithms (k-NN, SVM, clustering). Two techniques:

- **Clipping (`.clip()`)** — hard bounds, fast, transparent
- **Percentile capping** — data-driven, preserves the distribution's natural tail while removing extreme outliers

> **Instructor Note:** Clipping vs. dropping is a judgment call. For predictive maintenance, an extreme response time might be exactly the signal you want to learn from — you'd want to keep it but maybe log-transform it. For anomaly detection, you might want to keep it as a label rather than a feature. Always think about what the model will do with the value.

````python
# --- Fix cpu_percent: clip to valid range [0, 100] ---

n_over_100 = (df['cpu_percent'] > 100).sum()
n_under_0  = (df['cpu_percent'] < 0).sum()
print(f"cpu_percent values > 100: {n_over_100}")
print(f"cpu_percent values < 0:   {n_under_0}")

# .clip() is the cleanest way to enforce hard bounds
df['cpu_percent'] = df['cpu_percent'].clip(lower=0, upper=100)

print(f"\nAfter clip(0, 100):")
print(f"  Min: {df['cpu_percent'].min():.2f}")
print(f"  Max: {df['cpu_percent'].max():.2f}")
````

````python
# --- Fix memory_mb: negative values are invalid sensor reads ---

n_negative_mem = (df['memory_mb'] < 0).sum()
print(f"memory_mb negative values: {n_negative_mem}")

# Set negatives to NaN, then re-fill with median
# (We already computed mem_median above — reuse it)
df.loc[df['memory_mb'] < 0, 'memory_mb'] = np.nan
df['memory_mb'] = df['memory_mb'].fillna(mem_median)

print(f"After fix — Min: {df['memory_mb'].min():.0f} MB, Max: {df['memory_mb'].max():.0f} MB")
````

````python
# --- Fix response_time_ms: cap at 99th percentile ---

# Plot distribution BEFORE capping
fig, axes = plt.subplots(1, 2, figsize=(14, 4))

axes[0].hist(df['response_time_ms'], bins=60, color='steelblue', edgecolor='white')
axes[0].set_title('response_time_ms — BEFORE capping')
axes[0].set_xlabel('Response Time (ms)')
axes[0].set_ylabel('Frequency')
axes[0].axvline(df['response_time_ms'].quantile(0.99), color='red',
                linestyle='--', label='99th percentile')
axes[0].legend()

# Compute the cap threshold
p99 = df['response_time_ms'].quantile(0.99)
print(f"99th percentile of response_time_ms: {p99:.1f} ms")
print(f"Values above p99: {(df['response_time_ms'] > p99).sum()}")

# Cap at p99
df['response_time_ms'] = df['response_time_ms'].clip(upper=p99)

# Plot distribution AFTER capping
axes[1].hist(df['response_time_ms'], bins=60, color='#4caf50', edgecolor='white')
axes[1].set_title('response_time_ms — AFTER capping at 99th pct')
axes[1].set_xlabel('Response Time (ms)')
axes[1].set_ylabel('Frequency')

plt.tight_layout()
plt.show()

print(f"\nAfter cap — Max: {df['response_time_ms'].max():.1f} ms  (was {p99:.0f} ms threshold)")
````

## Step 7: Parse Log Messages with Regex

The `message` field is a goldmine of structured information trapped in an unstructured string. Natural language processing approaches (which we cover in Module 4) can extract semantic meaning, but for well-formatted log files, **regex is faster, more reliable, and fully explainable**.

We want to extract two new features:

1. **`extracted_error_code`** — the `ERR-XXXX` or `NX-XXXX` code embedded in the message text
2. **`operation_type`** — whether the event involved a READ, WRITE, DELETE, or UPDATE

Why extract these separately?
- They're categorical features a model can directly use
- Raw message strings are high-cardinality text that most models can't consume directly
- `error_code` in the dedicated column has ~30% missing — we may recover values from the message

> **Instructor Note:** This is a preview of feature engineering from unstructured text. In Module 4 we'll use transformer embeddings to extract semantics from messages that don't have predictable patterns. Regex and embeddings are complementary, not competing.

````python
# --- Extract error code from message ---

# Regex pattern explanation:
#   (ERR|NX)   — match either 'ERR' or 'NX' prefix
#   -          — literal hyphen
#   \d{3,4}    — 3 to 4 consecutive digits
error_code_pattern = r'((?:ERR|NX)-\d{3,4})'

# .str.extract() returns a DataFrame; [0] gets the first capture group
df['extracted_error_code'] = df['message'].str.extract(error_code_pattern, expand=False)

print("Error code extraction:")
print(f"  Extracted from message:    {df['extracted_error_code'].notna().sum()} rows")
print(f"  Original error_code col:   {df['error_code'].notna().sum()} rows non-null")
print()
print("Sample extractions:")
sample_mask = df['extracted_error_code'].notna()
print(df.loc[sample_mask, ['message', 'error_code', 'extracted_error_code']].head(6).to_string(index=False))
````

````python
# --- Reconcile: fill missing error_code from message extraction ---

# Where the dedicated column is NaN but we extracted a code from the message, use that
recovered_mask = df['error_code'].isna() & df['extracted_error_code'].notna()
df.loc[recovered_mask, 'error_code'] = df.loc[recovered_mask, 'extracted_error_code']

print(f"Recovered {recovered_mask.sum()} error codes from message text.")
print(f"error_code null count: {df['error_code'].isna().sum()} (was higher before)")
````

````python
# --- Extract operation type from message ---

# Pattern: find the first occurrence of READ, WRITE, DELETE, or UPDATE (word boundary, case-insensitive)
operation_pattern = r'\b(READ|WRITE|DELETE|UPDATE)\b'

df['operation_type'] = df['message'].str.extract(operation_pattern, expand=False, flags=re.IGNORECASE)

# Normalize to uppercase
df['operation_type'] = df['operation_type'].str.upper()

print("Operation type distribution:")
print(df['operation_type'].value_counts(dropna=False).to_string())
print()
print(f"Rows with no operation extracted: {df['operation_type'].isna().sum()}")
````

````python
# Visualize extracted features
fig, axes = plt.subplots(1, 2, figsize=(13, 4))

# Operation type breakdown
op_counts = df['operation_type'].value_counts()
colors_op  = ['#2196f3', '#4caf50', '#f44336', '#ff9800']
axes[0].bar(op_counts.index, op_counts.values, color=colors_op[:len(op_counts)])
axes[0].set_title('Extracted Operation Types')
axes[0].set_ylabel('Count')
for i, (idx, val) in enumerate(op_counts.items()):
    axes[0].text(i, val + 1, str(val), ha='center', va='bottom', fontsize=9)

# Error code frequency
ec_counts = df['error_code'].value_counts().head(8)
axes[1].barh(ec_counts.index, ec_counts.values, color='#9c27b0')
axes[1].set_title('Top Error Codes (after recovery)')
axes[1].set_xlabel('Count')
axes[1].invert_yaxis()

plt.tight_layout()
plt.show()
````

## Key Takeaways

### Techniques Covered

- **`pd.to_datetime(errors='coerce')`** — gracefully parses messy timestamp strings; invalid values become `NaT` so you can handle them deliberately rather than crashing your pipeline

- **`.str.strip()` + `.str.upper()` + `.replace(map)`** — a three-step pattern for standardizing categorical columns; always apply in this order

- **Median imputation for skewed data** — skewed distributions (response times, CPU under heavy load) have means pulled by extreme values; median is a more robust central estimate

- **Domain-constant imputation** — `disk_io = 0 when missing` is only valid because we understand what "missing" means in this specific system; this requires domain knowledge, not just statistics

- **`.clip(lower, upper)`** — the fastest way to enforce hard physical bounds on sensor readings

- **Percentile capping (99th pct)** — a data-driven alternative to hard-coded bounds; adapts to the actual distribution of your data

- **`.str.extract(regex)` for feature engineering** — extracting structured fields from unstructured log text is often the highest-ROI step in an infrastructure ML pipeline

- **Cross-column imputation (error_code recovery)** — when one column has missing values, check whether another column (here: `message`) contains the same information in a different form

### The Golden Rule

> **Every cleaning decision is a modeling decision. Document your choices, the assumptions they rest on, and what could break them.**
